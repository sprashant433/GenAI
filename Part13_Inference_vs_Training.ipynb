### 1. **Inference vs. Training** – The Restaurant Analogy

Think of an **LLM project like running a restaurant**:

* **Inference** = Serving dishes to customers on demand (real-time use).
* **Training** = Teaching your chefs new recipes from scratch.
* **Fine-tuning** = Teaching your chefs slight variations of existing recipes to fit your restaurant style.

We’ve mostly been working on **Inference** (improving answers at runtime without changing the chef’s core skills).

---

### 2. **Techniques to Improve Inference**

*(“Ways to spice up the dish before it leaves the kitchen”)*

| Technique                                | Analogy                                                            | Pros                              | Cons                        | Example                                |
| ---------------------------------------- | ------------------------------------------------------------------ | --------------------------------- | --------------------------- | -------------------------------------- |
| **Multi-shot Prompting**                 | Give the chef multiple example dishes before asking for a new one  | Fast to try, no retraining needed | Limited by context window   | Few-shot prompt for sentiment analysis |
| **Prompt Chaining**                      | Break big orders into smaller courses and serve in sequence        | Keeps complexity manageable       | More calls = more latency   | Step-by-step question answering        |
| **Tools / Function Calling**             | Give the chef a recipe book or calculator mid-service              | Accesses up-to-date info          | Requires integration        | LLM calls a weather API                |
| **RAG (Retrieval-Augmented Generation)** | Chef checks the restaurant’s private recipe archive before cooking | Accurate without retraining       | Needs knowledge base, infra | Searching a private product catalog    |

---

### 3. **Training from Scratch? Usually Not.**

* Training a multi-billion parameter model **from scratch** = Building your own restaurant chain, hiring all chefs, growing your own vegetables → **\$10M–\$100M**.
* Instead → **Transfer Learning**:
  Use a **pre-trained chef** and teach them your special dishes (**fine-tuning**).

---

### 4. **Finding Training Data** – "Where do we get our ingredients?"

* **Your own proprietary data** – best quality for your domain.
* **Kaggle** – competitive public datasets.
* **HuggingFace Datasets** – wide variety of open datasets.
* **Synthetic data** – generated via GPT/LLM.
* **Specialist companies (e.g., scale.com)** – curated data collection.

---

### 5. **Data Preparation** – "Clean the kitchen before cooking"

1. **Investigate** – What’s in the dataset?
2. **Parse** – Convert to usable format.
3. **Visualize** – Spot trends/anomalies.
4. **Assess Quality** – Are there errors, missing values?
5. **Curate** – Remove noise, keep the good stuff.
6. **Save** – Version control data (Parquet, Delta, etc.).

*Example*: Use Pandas to profile sales data, detect outliers, then save clean dataset in GCS.

---

### 6. **Evaluating Performance** – "Taste Test"

Two perspectives:

#### **Model-centric metrics (technical)**

* **Training Loss** – How wrong the model is during training.
* **Validation Loss** – How wrong it is on unseen data.
* **RMSLE** – Penalizes large under/over-estimations.

#### **Business-centric metrics**

* **Average Price Difference**
* **% Price Difference**
* **% Estimates that are “good”** (meets business threshold).

*Example*: Predicting house prices – RMSLE improves from `0.42` → `0.30`, business metric improves from 60% → 80% good predictions.

### 7. **5-Step Strategy for Applying LLM to a Commercial Problem**

#### **1. Understand** – "Know what the restaurant serves"

* Gather business requirements.
* Define **business-centric metrics** first.
* Understand data (quantity, quality, format).
* Define constraints:

  * Cost 💰
  * Scalability 🏗️
  * Latency ⏱️
  * Timeline 📅

---

#### **2. Prepare** – "Get your ingredients ready"

* Research **existing solutions** (including non-LLM).
* Compare relevant LLMs:

  * Context length, cost, license
  * Benchmarks (MMLU, HELM)
  * Leaderboards (HuggingFace Open LLM Leaderboard)
  * Specialist scores (e.g., for legal QA).
* Clean, preprocess, split data.

---

#### **3. Select** – "Pick your chef"

* Choose candidate models.
* Run experiments.
* Train/validate with curated data.

---

#### **4. Customize** – "Teach your chef your recipes"

| Method          | Pros                       | Cons                                 | Example Use Case                    |
| --------------- | -------------------------- | ------------------------------------ | ----------------------------------- |
| **Prompting**   | Cheap, quick               | Context-limited, diminishing returns | Customer support bot                |
| **RAG**         | Low data need, scalable    | Harder infra, needs accurate KB      | Product catalog Q\&A                |
| **Fine-tuning** | Deep expertise, tone/style | Expensive, risk of forgetting        | Domain-specific contract summarizer |

---

#### **5. Productionize** – "Run the kitchen"

* Define API between LLM and platform.
* Choose hosting (Vertex AI, AWS Sagemaker, vLLM on GCP).
* Handle scaling, monitoring, compliance.
* Continuously retrain & monitor metrics.

---

### 📌 Key to remember

* **Prompting is like seasoning** → quick to change, but can’t make a bad dish good.
* **RAG is like consulting a cookbook mid-service** → helps with accuracy, but only if the book is up-to-date.
* **Fine-tuning is like sending your chef to culinary school** → costly but results in lasting skills.
* **Always measure success in business terms first** – RMSLE of `0.20` means nothing if the customer still gets the wrong pizza.
