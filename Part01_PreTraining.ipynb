# STAGE 1: PRETRAINING

Pretraining is the foundational step where a model learns the statistical structure of human language from massive amounts of raw text data.

---

> 💡 **Key Insight:**

- The model never sees individual letters; it sees token IDs.

## 1.2 — Tokenization
---
**Tokens are the building blocks of text that LLMs understand.**

🔗 **Tools:**

* [OpenAI's Tokenizer](https://platform.openai.com/tokenizer)
* [Tik Tokenizer](https://tiktokenizer.vercel.app/)

**What happens:**
- Text is converted to tokens using algorithms like **Byte-Pair Encoding (BPE)**.
- Pipeline: **raw text → binary → bytes → tokens (BPE)**

Computers don’t understand text. They understand **numbers (binary)**.

#### Example:

```text
"cat"
```

Each character is mapped to a number using encoding (like UTF-8):

| Character | ASCII/Unicode |
| --------- | ------------- |
| c         | 99            |
| a         | 97            |
| t         | 116           |


### ⚡ 2. Numbers → Binary (0s and 1s)

Each number is converted into **binary (base-2)**.

Example:

```text
c → 99 → 01100011
a → 97 → 01100001
t → 116 → 01110100
```

👉 Now we have:

```text
"cat" → 01100011 01100001 01110100
```

### 📦 3. Binary → Bytes (8 bits = 1 byte)

👉 A **byte = 8 bits**

Each of the above chunks is already **1 byte**:

```text
01100011 → 1 byte (c)
01100001 → 1 byte (a)
01110100 → 1 byte (t)
```

So:

```text
"cat" = 3 bytes
```

💡 Important:

* English characters → usually 1 byte each
* Emoji / other languages → multiple bytes

Example:

```text
"😊" → 4 bytes in UTF-8
```

### 🧠 4. Bytes → Tokens (This is where BPE comes in)

Now comes the LLM-specific part 👇

#### Key idea:

👉 Models **don’t read raw bytes directly**
👉 They convert bytes → **tokens**


#### 🔹 Step 4.1: Start from Bytes

From earlier:

```text
"cat" → [99, 97, 116]
```

#### 🔹 Step 4.2: Build Vocabulary using BPE

BPE learns patterns like:

```text
"c" + "a" → "ca"
"ca" + "t" → "cat"
```

#### 🔹 Step 4.3: Tokenization

Depending on vocabulary:

##### Case 1 (word exists):

```text
"cat" → ["cat"]
```

##### Case 2 (not in vocab):

```text
"caterpillar" → ["cat", "er", "pillar"]
```
---

### 🔄 Full Pipeline

```text
Text:        "cat"
↓
Characters:  ['c','a','t']
↓
Encoding:    [99, 97, 116]
↓
Binary:      01100011 01100001 01110100
↓
Bytes:       [byte1, byte2, byte3]
↓
BPE:         ["cat"] or ["ca","t"]
↓
Token IDs:   [1234]   (final input to model)
```

### 🤯 Why Not Use Raw Bytes Directly?

| Approach           | Problem             |
| ------------------ | ------------------- |
| Raw bytes          | No semantic meaning |
| Characters         | Too long sequences  |
| Words              | Huge vocab + OOV    |
| **BPE (subwords)** | ✅ Best tradeoff     |


### Important facts:
---

* **Modern tokenizers (like GPT) use Byte-level BPE**

  * Start from **bytes (0–255)**
  * Merge into **subwords (tokens)**
  * Even unusual inputs are tokenizable:

    ```text
    "@#$%^&*"
    ```

* **One token ≠ one word**

  * Tokens are **subword units**
  * Example:

    ```text
    "unbelievable"
    → ["un", "believ", "able"]
    ```

* **Token count affects:**

  * Cost 💰
  * Latency ⚡
  * Context window 📏

* **Models see tokens, not characters**

  * Struggle with:

    * Spelling tasks
    * Precise arithmetic
  * Example:

    ```text
    9.9  → split differently
    9.11 → split differently
    ```

* **What a tokenizer does:**

  * Splits text → **tokens (subwords)**
  * Maps tokens → **Token IDs (numbers)**
  * This mapping is just a **lookup table (no meaning yet)**

* **Vocabulary details:**

  * Typically **50,000 – 100,000 tokens**

  * Tokens are:

    * ❌ Not individual letters
    * ❌ Not full words
    * ✅ Subword units

  * Example:

    ```text
    "Rayleigh scattering"
    → ["Ray", "leigh", " sc", "attering"]
    ```

* **Final flow:**

  * Text → Tokens → Token IDs → Model → Next token prediction 🚀

---

**Code snippet:**

````python
from transformers import AutoTokenizer

text = "I love cats!"

tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased') # 'lapp0/Qwen2.5-1.5B-fixed-tokenizer'

tokenizer.get_vocab() # {'⁵': 1539,'uranium': 14247,'irresistible': 27149, ...}

tokens = tokenizer.tokenize(text) # ['i', 'love', 'cats', '!']

tokenizer.encode(text) # [101, 1045, 2293, 8870, 999, 102]

tokenizer.encode(text, add_special_tokens=False) # [1045, 2293, 8870, 999]

tokenizer.convert_tokens_to_ids(tokens) # [1045, 2293, 8870, 999]

tokenizer.decode([101, 1045, 2293, 8870, 999, 102]) # [CLS] i love cats! [SEP]'

tokenizer.convert_ids_to_tokens([101, 1045, 2293, 8870, 999, 102]) # ['[CLS]', 'i', 'love', 'cats', '!', '[SEP]']
````

## 1.3 — Neural Network Architecture: The Transformer

🔗 **Link:**

* [Transformer Architecture](https://bbycroft.net/llm)

The backbone of all modern LLMs is the **Transformer** (Vaswani et al., 2017).

### Architecture Overview:

```
Input Tokens → Embedding Layer → [N × Transformer Blocks] → LM Head → Next Token Probabilities

Each Transformer Block:
  ├── Multi-Head Self-Attention
  ├── Add & LayerNorm
  ├── Feed-Forward Network (MLP)
  └── Add & LayerNorm
```

### Key Parameters:

| Model | Parameters | Layers | Heads | d_model |
|-------|-----------|--------|-------|--------|
| GPT-2 Small | 117M | 12 | 12 | 768 |
| GPT-2 XL | 1.5B | 48 | 25 | 1600 |
| GPT-3 | 175B | 96 | 96 | 12288 |
| LLaMA-2 70B | 70B | 80 | 64 | 8192 |

|--------|-------------|-----------|
| **Purpose** | Convert text → numbers (IDs) | Convert IDs → meaningful vectors |
| **Input** | Raw text | Token IDs |
| **Output** | Integer IDs | Dense vectors (arrays of floats) |
| **When** | First step | Second step |

## 1.4 — Next-Token Prediction (The Pretraining Objective)

The model is trained with a **simple but powerful objective**: predict the next token given all previous tokens.

```
Input:   ["The", "cat", "sat", "on", "the"]
Target:  ["cat", "sat", "on", "the", "mat"]
```

This is **self-supervised learning** — no labels needed! The labels come from the text itself.

**Loss function:**
$$\mathcal{L} = -\sum_{t=1}^{T} \log P(x_t | x_1, x_2, ..., x_{t-1})$$

After pretraining, the model is a **Base Model** — also called an *"internet document simulator"*:
- Given the beginning of a document, it completes it in a plausible style.
- It is NOT yet an assistant — it just continues text.
