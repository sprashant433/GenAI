
# 📚 **Comparing LLMs**

---

## **1. Start with the Basics**

When you compare Large Language Models (LLMs), think of it like **buying a car** 🚗 — you don’t just ask “how fast is it?”; you look at horsepower, fuel type, safety ratings, price, and comfort.

### **Key Specs to Check:**

* **Parameters** – The number of “neurons” in the model’s brain 🧠.
  Example: GPT-3 has 175B parameters, LLaMA 2 comes in 7B, 13B, and 70B sizes.
  Analogy: More parameters = a bigger brain, but not always a *smarter* brain if it wasn’t trained well.

* **Context Length** – How much information it can remember from the conversation.
  Example: GPT-4 Turbo supports \~128k tokens, so it can handle entire books.
  Analogy: Like the size of a whiteboard — the bigger it is, the more notes you can write before erasing.

* **Pricing** – Either **API cost** (e.g., \$0.002 per 1K tokens) or **subscription**.
  Example: OpenAI’s GPT-4 Turbo is cheaper per token than GPT-4 base.

## **2. Extended Basics (The “Spec Sheet”)**

### **A. Open-source vs Closed-source**

* **Open-source**: You can download and run locally (e.g., LLaMA, Mistral).
* **Closed-source**: Hosted only by the vendor (e.g., GPT-4, Claude).
  Analogy: Open-source is like building your own PC; closed-source is like buying a sealed Mac.

### **B. Release Date & Knowledge Cutoff**

* Release date tells you **how fresh** the model is.
* Knowledge cutoff tells you **what it knows** by default.
  Example: GPT-4 cutoff = Dec 2023 (knows events up to then).

### **C. Parameters & Training Tokens**

* **Parameters**: Size of the brain.
* **Training tokens**: Amount of text data it read during training.
  Analogy: You might have a large brain (parameters), but if you never read books (training tokens), you won’t know much.

### **D. Inference Cost**

* Cost to run the model in production.
  Example: Running LLaMA-70B locally requires expensive GPUs.

### **E. Training & Build Cost**

* Training = Pretraining from scratch (millions of dollars for frontier models).
* Build = Fine-tuning or RAG pipeline setup cost.

### **F. Time to Market**

* How fast you can go from “idea” → “working product”.

### **G. Rate Limits, Speed, and Latency**

* Rate limits: Maximum requests per minute/hour.
* Speed: Tokens per second generated.
* Latency: Delay before first token appears.

### **H. License**

* Dictates usage rights.
  Example: LLaMA 2 license allows commercial use; GPT-4 license doesn’t let you run it locally.

## **3. 📏 Chinchilla Scaling Law**

This is like **nutrition for the brain** analogy — balance between **brain size** (parameters) and **food** (training data).

* **Law:**

  * Number of parameters ≈ Number of training tokens for optimal performance.
  * If you double parameters, you also need \~double training tokens to keep improving.
* If you overfeed a small brain (too much data for small parameters), waste of food.
* If you underfeed a big brain (too few tokens for large parameters), it’s like a bodybuilder eating only a salad.

## **4. 🏁 Common Benchmarks (General Intelligence)**

Benchmarks = **exams for LLMs** 📜.

* **ARC** – AI2 Reasoning Challenge (science reasoning questions).
* **DROP** – Discrete Reasoning Over Paragraphs (logic & math from text).
* **HellaSwag** – Commonsense reasoning for everyday scenarios.
* **MMLU** – Massive Multitask Language Understanding (college-level exams in many subjects).
* **TruthfulQA** – Checks if the model gives truthful answers (avoiding misinformation).
* **Winogrande** – Grammar + commonsense pronoun resolution.
* **GSM8K** – Grade school math word problems.

---

## **5. 🎯 Specific Benchmarks**

* **ELO** – Rating from human head-to-head votes (like chess ratings).
* **HumanEval** – Code generation benchmark in Python.
* **MultiPL-E** – Multi-language code generation evaluation.

---

## **6. 🚧 Limitations of Benchmarks**

* Inconsistent usage.
* Narrow scope — doesn’t cover all skills.
* Misses nuanced reasoning.
* Risk of **training data leakage** (model saw the test in training).
* Overfitting to benchmarks.
* **Emerging concern:** Models recognizing they are being tested and gaming results.

---

## **7. 🔥 Hard, Next-Level Benchmarks**

* **GPQA** – Google Proof Question and Answer (advanced proof-based Q\&A).
* **BBHard** – Big-Bench Hard (very tough general tasks).
* **Math Lv 5** – Graduate-level math.
* **IFEval** – Instruction Following Evaluation.
* **MuSR** – Multistep Soft Reasoning (multi-hop reasoning).
* **MMLU-Pro** – More difficult version of MMLU.

Analogy: If normal benchmarks are “high school finals,” these are **PhD qualifying exams**.

👉 **“Grab Big Meals In My Mug.”**

* **Grab** → GPQA
* **Big** → BBHard
* **Meals** → Math Lv 5
* **In** → IFEval
* **My** → MuSR
* **Mug** → MMLU-Pro

## **9. ⚔ [The Arena](https://lmarena.ai/leaderboard)**

* Pits two LLMs against each other.
* Humans vote on which output is better (blind test).
* Generates **ELO ratings**.
* Open to public participation.
