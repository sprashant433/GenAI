## AI Agents Fundamentals

### What are AI Agents?

**Layman Definition:**
Imagine a personal assistant that doesn't just answer your questions, but actually takes actions and makes decisions to complete tasks. That's an AI agent - it's like the difference between asking someone "What's the weather?" (just getting an answer) versus "Handle my schedule today" (taking actions).

**Technical Definition:**
AI Agents are autonomous programs where **LLM outputs control the workflow execution** rather than simply providing text responses. The LLM's decisions determine what happens next in the application flow.

### Key Characteristics (one or more of these):

**1. Multiple LLM Calls**
- **Layman:** Like consulting an expert multiple times for different parts of a big project
- **Technical:** Task decomposition where each LLM call handles a specific subtask (e.g., research → analyze → synthesize)
- **Example:** Writing an essay might involve: outline generation → section writing → fact-checking → editing

**2. Tool Usage**
- **Layman:** The AI can use "tools" like calculators, search engines, or databases - just like you use apps on your phone
- **Technical:** LLMs invoke external functions, APIs, or systems through structured function calling
- **Example:** Calling a weather API, querying a database, sending an email

**3. Interactive Environment**
- **Layman:** The AI can "look around" and interact with its surroundings, like a robot exploring a room
- **Technical:** Agent perceives state changes in its environment and adapts behavior accordingly (observe → act → observe cycle)
- **Example:** Monitoring file system changes, reading user inputs, checking task completion status

**4. Planning Capabilities**
- **Layman:** The AI can create a plan and coordinate multiple steps to reach a goal
- **Technical:** High-level task decomposition with dependency management and sequencing logic
- **Example:** Planning a research project: identify sources → gather data → analyze → write report

**5. Autonomy**
- **Layman:** The AI can work independently without asking for help at every step
- **Technical:** Self-directed goal-seeking behavior with minimal human intervention, using feedback loops to adjust strategies
- **Example:** Debugging code by trying different solutions until tests pass

### The Fundamental Shift:
**Traditional LLM App:** User → Prompt → LLM → Response → Done  
**Agent:** User → Goal → LLM decides actions → Execute → LLM evaluates → Repeat until goal achieved

## What Makes an AI "Agentic"?

| Characteristic | Description |
|---------------|-------------|
| **Goal-directed** | Works toward an objective, not just one response |
| **Tool use** | Can call external APIs, run code, search web |
| **Memory** | Maintains context across steps |
| **Planning** | Decides what to do next dynamically |
| **Action** | Takes real-world actions (send email, book meeting) |

### Fundamental Shift
- **LLM (one-shot):** Input → Output
- **Agent (loop):** Input → Think → Act → Observe → Think → Act → ... → Output

### The Agent Loop

![Agent Loop](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_agent_loop.png)

The agent continuously:
1. Receives input from **Human**
2. Makes an **LLM Call** (think + decide)
3. Takes an **Action** on the **Environment** (tool call, API, file system)
4. Gets **Feedback** from the environment
5. Repeats until a **STOP** condition is met

## Agentic Systems: Workflows vs Agents

### Anthropic's Distinction:

**The Core Question:** Who's in control - your code or the LLM?

### 1. Workflows

**Layman:** Like following a recipe - every step is written down beforehand, and you follow it exactly every time.

**Technical Characteristics:**
- **Predefined code paths** - Deterministic execution with if-then logic
- LLMs + tools orchestrated through fixed, programmer-defined logic
- Control flow determined at design time, not runtime
- Predictable, controllable, easier to debug and test

**When to Use Workflows:**
- Task steps are known and consistent
- Need predictable costs and execution time
- Regulatory/compliance requirements demand auditability
- Quality control is critical

**Example Workflow:**
```
1. Extract text from PDF
2. Summarize with LLM
3. Translate summary
4. Save to database
```
*Each step always happens in this order*

---

### 2. Agents

**Layman:** Like hiring a consultant who figures out what to do as they go - they assess the situation and decide the best approach on the fly.

**Technical Characteristics:**
- **Dynamic control** - LLM decides next actions at runtime
- Self-directed processes with adaptive tool usage
- Maintains control over task accomplishment strategy
- Non-deterministic execution paths

**When to Use Agents:**
- Tasks require flexible problem-solving
- Unknown or varying solution paths
- Need to handle unexpected situations
- Creative or exploratory tasks

**Example Agent:**
```
Goal: "Debug why the website is slow"
Agent decides dynamically:
→ Check server logs?
→ Run performance tests?
→ Query database metrics?
→ Review recent code changes?

(Path depends on what it discovers)
```

---

### Key Difference Illustrated:

**Workflow = GPS navigation:**
- "Turn left in 500 feet, then turn right"
- Fixed route calculated upfront
- Efficient and predictable

**Agent = Human driver:**
- Sees traffic, decides to take alternate route
- Adapts to conditions in real-time
- Flexible but less predictable

**Pro Tip:** Start with workflows! Add agent behavior only when you truly need the flexibility. Most problems can be solved with workflows, which are easier to build, test, and maintain.

### Pattern 1. Prompt Chaining

**Layman:** Like an assembly line - each worker does one specific job and passes the result to the next worker. Each step specializes in one thing and does it well.

**Technical Concept:**
Task decomposition into sequential sub-tasks where each LLM call receives the output from the previous call as input. Each step uses specialized prompts optimized for that specific subtask.

**How it Works:**
```
Input → [LLM Call 1] → Output 1 → [LLM Call 2] → Output 2 → [LLM Call 3] → Final Output
```

**Real Example - Document Processing:**
```
Raw Text → [Extract Key Points] → Key Points →
[Organize by Theme] → Themed Summary →
[Write Executive Summary] → Final Report
```

**When to Use:**
- Tasks have clear sequential steps
- Each step needs different prompting strategies
- Want to maintain focus at each stage
- Need intermediate outputs for debugging

**Benefits:**
- **Modularity:** Easy to modify individual steps
- **Clarity:** Each step has a single responsibility
- **Debuggability:** Can inspect intermediate outputs
- **Optimization:** Different models/settings per step

**Trade-offs:**
- Slower than parallelization (sequential)
- Errors propagate through the chain
- Requires careful prompt engineering per step

![Pipeline](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_pipeline.png)


**IN → LLM1 → Gate → LLM2 → LLM3 → OUT**

Each LLM call's output feeds the next. A **Gate** can add conditional logic — only proceed if quality passes.

````python
from openai import OpenAI
client = OpenAI()

def chain_prompts(initial_input: str) -> str:
    """Sequential LLM calls — each output feeds the next."""

    # Step 1: Extract key business requirements
    step1 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "Extract the key business requirements as bullet points."},
            {"role": "user",   "content": initial_input}
        ]
    ).choices[0].message.content

    # Step 2: Design a solution from requirements
    step2 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "Design a technical solution for these requirements."},
            {"role": "user",   "content": step1}
        ]
    ).choices[0].message.content

    # Step 3: Write implementation plan from solution
    step3 = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "Write a detailed implementation plan."},
            {"role": "user",   "content": step2}
        ]
    ).choices[0].message.content

    return step3
````

### Pattern 2. Routing

**Layman:** Like a receptionist directing calls - different types of questions go to different specialized departments. You wouldn't ask the sales team about technical bugs!

**Technical Concept:**
Input classification followed by conditional execution paths. A router (often an LLM or rule-based classifier) determines which specialized sub-task should handle the input.

**How it Works:**
```
Input → [Router/Classifier] → Route Decision
                               ↓
            ┌──────────────────┼──────────────────┐
            ↓                  ↓                   ↓
    [Handler A]         [Handler B]         [Handler C]
            ↓                  ↓                   ↓
         Output             Output              Output
```

**Real Example - Customer Support:**
```
Customer Query → [Classify Intent]
                        ↓
    ┌──────────────────┼──────────────────┐
    ↓                  ↓                   ↓
[Sales Team]    [Technical Support]   [Billing]
  (Pricing)         (How-to)          (Invoice)
```

**When to Use:**
- Multiple distinct task types requiring different handling
- Need specialized prompts/models for different categories
- Want to optimize per category (cost, speed, quality)
- Have clearly distinguishable input types

**Implementation Approaches:**
1. **LLM-based:** Use LLM to classify intent
2. **Rule-based:** Keyword matching, regex patterns
3. **ML Classifier:** Trained classification model
4. **Hybrid:** Rules for obvious cases, LLM for ambiguous

**Benefits:**
- **Specialization:** Each handler optimized for its domain
- **Efficiency:** Use appropriate resources per task type
- **Scalability:** Easy to add new categories
- **Cost Optimization:** Use cheaper models for simple routes

**Example Code Structure:**
```python
def route_query(query):
    intent = classifier.classify(query)
    
    if intent == "technical":
        return technical_handler(query)
    elif intent == "billing":
        return billing_handler(query)
    else:
        return general_handler(query)
```

![Router](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_router.png)

**IN → LLM Router → LLM1 / LLM2 / LLM3 → OUT**

An LLM (or rule) examines the input and routes it to the most appropriate specialist model.

````python
from openai import OpenAI
client = OpenAI()

# Specialist system prompts
SPECIALISTS = {
    "technical": "You are a technical expert. Answer technical questions about code and software.",
    "billing":   "You are a billing specialist. Handle payments, invoices, and subscription questions.",
    "general":   "You are a helpful general assistant.",
}

def route_and_answer(user_query: str) -> str:
    """Route query to the right specialist via an LLM router."""

    # Router LLM decides which specialist to use
    routing = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{
            "role": "user",
            "content": f"""Classify this query into exactly one category: technical, billing, or general.
Reply with only the category word.

Query: {user_query}"""
        }]
    ).choices[0].message.content.strip().lower()

    specialist = SPECIALISTS.get(routing, SPECIALISTS["general"])
    print(f"Routed to: {routing}")

    # Specialist LLM answers
    answer = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": specialist},
            {"role": "user",   "content": user_query}
        ]
    ).choices[0].message.content

    return answer

# print(route_and_answer("My invoice shows wrong amount"))
# print(route_and_answer("How do I implement a binary search tree?"))
````

### Pattern 3: Parallelization (Coordinator + Aggregator)

**Layman:** Like dividing a big project among team members who work simultaneously - everyone does their part at the same time, then you combine all the results at the end. Much faster than doing everything one-by-one!

**Technical Concept:**
Concurrent execution of independent sub-tasks with result aggregation. Tasks are distributed across multiple LLM calls that execute simultaneously, then results are merged or synthesized.

**How it Works:**
```
Input → Split into Sub-tasks
          ↓
    ┌─────┼─────┬─────┐
    ↓     ↓     ↓     ↓
[Task 1][Task 2][Task 3][Task 4]  ← All run concurrently
    ↓     ↓     ↓     ↓
    └─────┼─────┴─────┘
          ↓
    [Aggregate Results]
          ↓
      Final Output
```

**Real Example - Document Analysis:**
```
Analyze 10 Chapters of a Book

Sequential: 10 chapters × 30 seconds = 5 minutes
Parallel: max(30 seconds) = 30 seconds

Process:
Chapter 1 → [Analyze] ┐
Chapter 2 → [Analyze] ├→ [Combine Summaries] → Full Book Summary
Chapter 3 → [Analyze] ┘
... (all run at once)
```

**When to Use:**
- Sub-tasks are independent (no dependencies)
- Want to reduce latency/execution time
- Processing multiple similar items
- I/O-bound operations (waiting for API responses)

**Common Use Cases:**
1. **Multi-document processing:** Analyze multiple docs simultaneously
2. **Batch translation:** Translate multiple paragraphs at once
3. **Sentiment analysis:** Process multiple reviews in parallel
4. **Data extraction:** Extract info from multiple sources concurrently

**Benefits:**
- **Speed:** Dramatically reduced execution time
- **Throughput:** Process more items per unit time
- **Resource Utilization:** Better use of available API quota
- **Scalability:** Easy to scale to more items

**Aggregation Strategies:**
1. **Simple Concatenation:** Join all outputs
2. **Voting/Consensus:** Multiple LLMs vote on answer
3. **Synthesis:** Another LLM combines results
4. **Statistical:** Average scores, combine metrics

**Trade-offs:**
- Higher concurrent API usage (may hit rate limits)
- More complex error handling
- Results need aggregation logic
- May cost more (more parallel calls)

**Example Code:**
```python
import asyncio

async def process_documents(docs):
    # Create tasks for all documents
    tasks = [analyze_document(doc) for doc in docs]
    
    # Run all tasks concurrently
    results = await asyncio.gather(*tasks)
    
    # Aggregate results
    return synthesize_results(results)
```

![Parallelization](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_parallelization.png)

**IN → Coordinator → [LLM1 ‖ LLM2 ‖ LLM3] → Aggregator → OUT**

Multiple LLMs work simultaneously on different sub-tasks. Results are aggregated into a final answer.


````python
import asyncio
from openai import OpenAI
client = OpenAI()

async def parallel_analysis(topic: str) -> dict:
    """Run multiple analyses simultaneously, then aggregate."""

    async def analyze(perspective: str) -> str:
        response = await asyncio.to_thread(
            client.chat.completions.create,
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": f"Analyze from a {perspective} perspective in 2-3 sentences."},
                {"role": "user",   "content": topic}
            ]
        )
        return response.choices[0].message.content

    # Coordinator: launch all analyses in parallel
    technical, business, risk = await asyncio.gather(
        analyze("technical"),
        analyze("business"),
        analyze("risk"),
    )

    # Aggregator: synthesize results
    combined = f"Technical: {technical}\n\nBusiness: {business}\n\nRisk: {risk}"
    summary = await asyncio.to_thread(
        client.chat.completions.create,
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "Synthesize these perspectives into a concise executive summary."},
            {"role": "user",   "content": combined}
        ]
    )

    return {
        "technical": technical,
        "business":  business,
        "risk":      risk,
        "summary":   summary.choices[0].message.content
    }

# result = asyncio.run(parallel_analysis("Migrating a monolith to microservices"))
# print(result["summary"])
````

### Pattern 4: Orchestrator-Workers (Synthesizer)

**Layman:** Like a project manager with a team - the manager breaks down the big project into smaller tasks, assigns them to team members, and then puts all the pieces together into a final deliverable.

**Technical Concept:**
Hierarchical task decomposition with centralized coordination. An "orchestrator" LLM dynamically plans, delegates to "worker" LLMs, and synthesizes their outputs into a cohesive result.

**How it Works:**
```
Complex Task → [Orchestrator LLM]
                     ↓
          Plans & Decomposes Task
                     ↓
    ┌───────────────┼───────────────┐
    ↓               ↓                ↓
[Worker 1]     [Worker 2]      [Worker 3]
 Subtask A      Subtask B       Subtask C
    ↓               ↓                ↓
    └───────────────┼────────────────┘
                    ↓
          [Orchestrator LLM]
           Synthesizes Results
                    ↓
              Final Output
```

**Real Example - Research Paper Writing:**
```
Task: "Write a research paper on AI Safety"

Orchestrator Plans:
1. Literature review
2. Current challenges analysis  
3. Proposed solutions
4. Future directions

Orchestrator Delegates:
Worker A → Review 2020-2023 papers
Worker B → Analyze alignment challenges
Worker C → Summarize proposed techniques
Worker D → Identify research gaps

Orchestrator Synthesizes:
→ Combines all sections
→ Ensures consistency
→ Writes introduction/conclusion
→ Final paper
```

**When to Use:**
- Complex tasks requiring dynamic decomposition
- Sub-tasks aren't known upfront (need planning)
- Need centralized coordination and consistency
- Want to divide cognitive labor intelligently

**Variations:**

**1. Static Orchestration:**
- Orchestrator plans once at the start
- Workers execute predetermined tasks
- No dynamic replanning

**2. Dynamic Orchestration:**
- Orchestrator can replan based on worker outputs
- Adaptive task allocation
- Feedback loops between orchestrator and workers

**3. Hierarchical Orchestration:**
- Multiple levels: Super-orchestrator → Orchestrators → Workers
- For extremely complex tasks

**Benefits:**
- **Scalability:** Handle complex multi-faceted tasks
- **Modularity:** Workers are independent, reusable
- **Flexibility:** Dynamic planning adapts to task needs
- **Quality:** Centralized synthesis ensures coherence

**Design Considerations:**
- **Orchestrator Prompt:** Needs strong planning capabilities
- **Worker Specialization:** Each worker optimized for its domain
- **Communication Protocol:** Clear input/output formats
- **Error Handling:** What if a worker fails?

**Example Orchestrator Prompt:**
```
You are a project orchestrator. Given a complex task:
1. Break it into 3-5 independent subtasks
2. Specify what each subtask should accomplish
3. Later, you'll synthesize results into a final output

Task: {user_task}
Output your plan as JSON with subtask descriptions.
```

![Orchestrator](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_orchestrator.png)

**IN → Orchestrator → [Worker LLM1 ‖ Worker LLM2 ‖ Worker LLM3] → Synthesizer → OUT**

The **Orchestrator** breaks the task into sub-tasks and assigns them to worker LLMs. The **Synthesizer** (same or different LLM) combines sub-results into a final output.

````python
import asyncio, json
from openai import OpenAI
client = OpenAI()

async def orchestrate_task(complex_task: str) -> str:
    """Orchestrator breaks task into sub-tasks, workers execute, synthesizer combines."""

    # Orchestrator: decompose task
    plan_response = await asyncio.to_thread(
        client.chat.completions.create,
        model="gpt-4o-mini",
        messages=[{
            "role": "user",
            "content": f"""Break this task into exactly 3 parallel sub-tasks.
Return JSON: {{"subtasks": ["subtask1", "subtask2", "subtask3"]}}

Task: {complex_task}"""
        }]
    )
    plan = json.loads(plan_response.choices[0].message.content)
    subtasks = plan["subtasks"]
    print(f"Orchestrator created {len(subtasks)} sub-tasks")

    # Workers: execute sub-tasks in parallel
    async def worker(subtask: str) -> str:
        resp = await asyncio.to_thread(
            client.chat.completions.create,
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "You are a specialist. Complete your assigned sub-task thoroughly."},
                {"role": "user",   "content": subtask}
            ]
        )
        return resp.choices[0].message.content

    results = await asyncio.gather(*[worker(st) for st in subtasks])

    # Synthesizer: combine worker results
    combined = "\n\n".join(f"Sub-task {i+1} result:\n{r}" for i, r in enumerate(results))
    synthesis = await asyncio.to_thread(
        client.chat.completions.create,
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "Synthesize these sub-task results into a cohesive final answer."},
            {"role": "user",   "content": f"Original task: {complex_task}\n\n{combined}"}
        ]
    )
    return synthesis.choices[0].message.content

# result = asyncio.run(orchestrate_task("Write a comprehensive market analysis for electric vehicles"))
# print(result)
````

### Pattern 5: Evaluation & Re-running (Generator-Evaluator)

**Layman:** Like writing an essay and having a teacher review it - you write a draft, get feedback on what's wrong, improve it, and repeat until it's good enough. The writer and critic work together in a loop.

**Technical Concept:**
Iterative refinement loop where one LLM generates output and another evaluates/critiques it. The process repeats until quality criteria are met or max iterations reached.

**How it Works:**
```
Input → [Generator LLM] → Draft Output
              ↓
        [Evaluator LLM] → Assessment
              ↓
        Good enough?
         ↓         ↓
        Yes       No → [Generator LLM] + Feedback → Improved Draft
         ↓                                               ↓
    Final Output                                  [Loop back to Evaluator]
```

**Real Example - Code Generation:**
```
Task: "Write a function to parse JSON"

Iteration 1:
Generator: [Writes code]
Evaluator: "Missing error handling for invalid JSON"

Iteration 2:
Generator: [Adds try-catch blocks]
Evaluator: "Good! But needs type hints"

Iteration 3:
Generator: [Adds type annotations]
Evaluator: "Perfect! Meets all criteria"
→ Output final code
```

**When to Use:**
- Quality assurance is critical
- Iterative improvement yields better results
- Need to meet specific criteria or standards
- First-attempt outputs are often insufficient

**Key Components:**

**1. Generator:**
- Creates initial output
- Incorporates feedback in subsequent iterations
- Prompt includes previous critiques

**2. Evaluator (Critic):**
- Assesses output against criteria
- Provides specific, actionable feedback
- Decides if output meets quality bar
- Often uses structured output (pass/fail + reasoning)

**3. Loop Control:**
- Max iterations (prevent infinite loops)
- Convergence criteria (when to stop)
- Cost management (API call limits)

**Evaluation Strategies:**

**A. Criteria-Based:**
```python
criteria = [
    "Is factually accurate",
    "Uses proper grammar",
    "Addresses all points in prompt",
    "Appropriate tone and style"
]
```

**B. Scoring:**
```python
score = evaluator.rate(output, scale=1-10)
if score >= 8:
    return output
```

**C. Comparison:**
```python
# Generate multiple options, pick best
options = [generator() for _ in range(3)]
best = evaluator.choose_best(options)
```

**Real-World Applications:**
1. **Code Generation:** Generate → Test → Fix bugs → Repeat
2. **Content Creation:** Write → Check quality → Refine → Repeat
3. **Translation:** Translate → Check accuracy → Improve → Repeat
4. **Data Extraction:** Extract → Validate → Correct → Repeat

**Benefits:**
- **Quality:** Multiple passes improve output quality
- **Reliability:** Catch errors before final output
- **Adaptability:** Can define custom quality criteria
- **Self-improvement:** System gets better through iteration

**Trade-offs:**
- **Cost:** Multiple LLM calls per task
- **Latency:** Takes longer than single-pass
- **Complexity:** Need careful loop control
- **Diminishing Returns:** Each iteration helps less

**Example Implementation:**
```python
async def generate_with_evaluation(prompt, max_iterations=3):
    output = None
    feedback = ""
    
    for i in range(max_iterations):
        # Generate (incorporate previous feedback)
        output = await generator(prompt + feedback)
        
        # Evaluate
        evaluation = await evaluator(output)
        
        if evaluation.approved:
            return output
        
        feedback = f"\n\nFeedback: {evaluation.critique}"
    
    return output  # Return best attempt if max iterations reached
```

**Also Called:**
- Critic Pattern
- Refinement Loop
- Self-Critique Pattern
- Iterative Improvement

![Evaluator](https://raw.githubusercontent.com/sprashant433/GenAI/main/images/pattern_evaluator.png)

**IN → LLM Generator → LLM Evaluator → (Accepted: OUT) | (Rejected: feedback → LLM Generator again)**

One LLM generates, another evaluates. If rejected, feedback is sent back to regenerate. This loop continues until quality threshold is met.


````python
from pydantic import BaseModel
from openai import OpenAI
client = OpenAI()

class EvalResult(BaseModel):
    score: int       # 1–10
    passed: bool
    feedback: str

def generate_with_quality_gate(task: str, threshold: int = 7, max_retries: int = 3) -> str:
    """Generator-Evaluator loop: retry until quality threshold is met."""

    for attempt in range(1, max_retries + 1):
        # Generator LLM
        content = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": task}]
        ).choices[0].message.content

        # Evaluator LLM (structured output)
        eval_resp = client.beta.chat.completions.parse(
            model="gpt-4o-mini",
            messages=[{
                "role": "user",
                "content": f"Rate this response quality 1-10.\nTask: {task}\nResponse: {content}"
            }],
            response_format=EvalResult
        )
        result = eval_resp.choices[0].message.parsed
        print(f"Attempt {attempt}: score={result.score}, passed={result.passed}")

        if result.passed and result.score >= threshold:
            return content  # Accepted

        # Rejected: incorporate feedback for next attempt
        task = f"{task}\n\nPrevious feedback: {result.feedback}. Please improve."

    return content  # Return best attempt after max retries
````

### Pattern 6: No-Framework Agentic Loop

You don't need a framework for agentic patterns. Direct API calls with a **Gradio UI** and **evaluation loop** demonstrate the core concepts.

The pattern below builds a personal AI chatbot with:
- A **generator** (OpenAI) that answers as a persona
- An **evaluator** (Gemini) that checks quality
- A **re-runner** that retries with feedback if quality fails

````python
import os
import gradio as gr
from openai import OpenAI
from pydantic import BaseModel
from dotenv import load_dotenv

load_dotenv(override=True)

openai_client = OpenAI()
gemini_client = OpenAI(
    api_key=os.getenv("GOOGLE_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

class Evaluation(BaseModel):
    is_acceptable: bool
    feedback: str

NAME = "Alex"
BACKGROUND = "Alex is a senior software engineer with 10 years of experience in Python and cloud systems."

system_prompt = f"""You are acting as {NAME}.
Answer questions about {NAME}'s career, skills, and experience.
Be professional and engaging.

## Background:
{BACKGROUND}
"""

def evaluate(reply: str, message: str, history: list) -> Evaluation:
    """Evaluator LLM (Gemini) checks if the reply is acceptable."""
    eval_prompt = f"""You evaluate if an AI agent's response is acceptable quality.
The agent is playing the role of {NAME} on their personal website.

Context: {BACKGROUND}

User message: {message}
Agent reply: {reply}

Is this reply acceptable? Provide structured feedback."""

    response = gemini_client.beta.chat.completions.parse(
        model="gemini-2.0-flash",
        messages=[{"role": "user", "content": eval_prompt}],
        response_format=Evaluation
    )
    return response.choices[0].message.parsed

def rerun(reply: str, message: str, history: list, feedback: str) -> str:
    """Regenerate with evaluator feedback injected into system prompt."""
    improved_system = system_prompt + f"""

## Quality control rejected your previous answer
Your attempted answer: {reply}
Reason for rejection: {feedback}
Please improve your response."""

    messages = [{"role": "system", "content": improved_system}] + history + [{"role": "user", "content": message}]
    response = openai_client.chat.completions.create(model="gpt-4o-mini", messages=messages)
    return response.choices[0].message.content

def chat(message: str, history: list) -> str:
    """Main chat function: generate → evaluate → rerun if needed."""
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]

    # Generator
    reply = openai_client.chat.completions.create(model="gpt-4o-mini", messages=messages).choices[0].message.content

    # Evaluator
    evaluation = evaluate(reply, message, history)

    if evaluation.is_acceptable:
        print("Passed quality check")
    else:
        print(f"Quality check failed: {evaluation.feedback}")
        reply = rerun(reply, message, history, evaluation.feedback)

    return reply

# gr.ChatInterface(chat, type="messages").launch()
````

## Pattern Selection Guide

**Quick Decision Tree:**

1. **Sequential steps needed?** → Prompt Chaining
2. **Different input types?** → Routing
3. **Independent tasks, need speed?** → Parallelization
4. **Complex task needing breakdown?** → Orchestrator-Worker
5. **Need quality refinement?** → Evaluator-Optimizer

**Combining Patterns:**
These patterns can be combined! For example:
- Orchestrator-Worker + Parallelization (workers run in parallel)
- Routing + Prompt Chaining (route to different chains)
- Parallelization + Evaluator-Optimizer (evaluate multiple options)

**Key Takeaway:**
These patterns provide **deterministic, controllable workflows** - you know exactly what will happen. They give you the benefits of AI while maintaining predictability, debuggability, and cost control.

## By Contrast: Agents (vs Workflows)

**Fundamental Difference:** In workflows, YOUR code decides what happens next. In agents, THE LLM decides what happens next.

---

### Key Characteristics of Agents:

### 1. Open-ended Execution

**Layman:** Like giving someone a goal without telling them how to achieve it. "Fix the bug" not "First check logs, then review code, then test the fix." They figure out the path themselves.

**Technical:**
- No predetermined solution path or execution graph
- LLM explores problem space dynamically
- Outcome and intermediate steps not specified upfront
- Search-based problem solving

**Example:**
```
Workflow: "Extract data from PDF → Process with regex → Save to DB"
Agent: "Get me the customer data from this document"
  (Agent figures out: PDF? Excel? Image? What extraction method?)
```

**Why Open-ended:**
- Real-world problems often don't have fixed solutions
- Context determines best approach
- New situations require adaptive strategies

---

### 2. Feedback Loops (Observe-Act-Reflect)

**Layman:** Like learning to ride a bike - you try something, see what happens, adjust, and try again. The agent learns from each action's result.

**Technical Concept:**
Continuous perception-action cycles where the agent:
1. **Observes:** Checks result of previous action
2. **Reflects:** Evaluates if goal is closer or further
3. **Adjusts:** Modifies strategy based on feedback
4. **Acts:** Takes next action
5. **Repeat** until goal achieved or max iterations

**The Feedback Loop Pattern:**
```
Goal → [Plan Action] → [Execute Action] → [Observe Result]
  ↑                                              ↓
  └──────────────── [Adjust Strategy] ←─────────┘
```

**Real Example - Debugging Code:**
```
Iteration 1:
Action: "Check error logs"
Result: "TypeError on line 45"
Reflection: "Need to examine line 45"

Iteration 2:
Action: "Read code at line 45"
Result: "Variable 'user' used but could be None"
Reflection: "Need to add None check"

Iteration 3:
Action: "Add if user is not None check"
Result: "Code runs, tests pass"
Reflection: "Goal achieved!"
```

**Types of Feedback:**
1. **Success/Failure:** Did the action work?
2. **Partial Progress:** Are we closer to the goal?
3. **Environmental Changes:** What changed in the environment?
4. **Error Messages:** What went wrong?
5. **Validation Results:** Does output meet criteria?

**Why Feedback Loops Matter:**
- Enable self-correction
- Allow recovery from errors
- Support learning from experience
- Handle unexpected situations

---

### 3. No Fixed Path (Dynamic Decision-Making)

**Layman:** Like a choose-your-own-adventure book where the next page isn't decided until you see what happened on the current page. The agent decides "what's next?" at each step.

**Technical:**
- Control flow determined at runtime, not design time
- LLM makes branching decisions based on current state
- Can backtrack, retry, or change approach mid-execution
- Responds to changing conditions and new information

**Workflow vs Agent Path:**

**Workflow (Fixed):**
```python
def workflow():
    step1()  # Always happens
    step2()  # Always happens  
    step3()  # Always happens
    return result
```

**Agent (Dynamic):**
```python
def agent():
    while not goal_achieved():
        current_state = observe()
        next_action = llm_decides(current_state, history)
        result = execute(next_action)
        
        if needs_different_approach(result):
            strategy = llm_plans_new_strategy()
    
    return final_result
```

**Real Example - Research Task:**

**Workflow Approach:**
```
1. Search Google
2. Read top 3 articles
3. Summarize findings
```

**Agent Approach:**
```
Goal: "Understand current state of quantum computing"

Agent's Dynamic Path:
→ Search quantum computing
→ Finds recent breakthrough paper
→ Realizes needs background on qubits
→ Searches "qubits explained"
→ Reads primer
→ Returns to breakthrough paper
→ Needs to understand error correction
→ Searches "quantum error correction"
→ Now has context to understand breakthrough
→ Summarizes findings

(Path emerged from discoveries made along the way)
```

**Why No Fixed Path:**
- Problem space too complex to plan upfront
- New information changes what's needed next
- Exploratory tasks have unknown requirements
- Creative problem-solving requires flexibility

### Visual Comparison:

**Workflow:**
```
A → B → C → D → Result
(Same path every time)
```

**Agent:**
```
Start → A → B → Hmm, need X first
              → X → B → C → Failed, try Y
                         → Y → Z → Result
(Different path based on observations)
```

## Agents: When the Benefits Outweigh the Risks

**Use Agents When:**

1. **Problem is exploratory**
   - "Find the cause of this performance issue"
   - Solution path unknown upfront

2. **Environment is dynamic**
   - Web scraping (pages change structure)
   - System troubleshooting (errors vary)

3. **Tasks require creativity**
   - "Improve this design"
   - "Propose solutions to X"

4. **Need adaptive behavior**
   - User needs vary by context
   - Different data requires different processing

5. **Workflow would be too complex**
   - Too many branches to code explicitly
   - Combinatorial explosion of paths

**Don't Use Agents When:**
- Path is well-defined (use workflows)
- Need predictable cost/time
- Compliance requires auditability
- Quality can't tolerate variance

---

## The Trade-off:

**Workflows:**
- ✅ Predictable, testable, controllable
- ❌ Rigid, can't adapt

**Agents:**
- ✅ Flexible, adaptive, creative
- ❌ Unpredictable, expensive, hard to debug

**Best Practice:** Start with workflows. Add agent behavior only where you truly need the adaptability.

## Risks of Agent Frameworks

**Reality Check:** Agents sound amazing, but they introduce real challenges. Here's what you need to know:

---

### Risk 1: Unpredictable Path

**The Problem:**

**Layman:** You don't know what the agent will do until it does it. It's like giving someone directions and they decide to take a completely different route - you can't debug what you can't predict.

**Technical:**
- Non-deterministic execution paths
- Same input can produce different action sequences
- Difficult to reproduce bugs (agent may not take same path twice)
- Hard to unit test (what do you test if path varies?)
- Debugging requires tracing actual execution, not reading code

**Real Example:**
```
Bug Report: "Agent failed to complete task"

Developer: "What did it do?"
User: "No idea, it just... didn't work"

Problem: Agent took 15 different actions across 3 minutes.
Which step failed? Why? Can't reproduce it.
```

**Why This Happens:**
- LLM responses have randomness (temperature > 0)
- Agent state affects decisions (history-dependent)
- External API results vary
- Timing issues (concurrent operations)

**Mitigation Strategies:**

1. **Comprehensive Logging:**
```python
# Log every decision point
logger.info(f"Agent decided: {action}")
logger.info(f"Reasoning: {llm_reasoning}")
logger.info(f"Current state: {state}")
logger.info(f"Result: {result}")
```

2. **Step-by-step Tracking:**
- Record each action taken
- Save LLM inputs/outputs
- Track state transitions
- Timestamp everything

3. **Replay Capability:**
```python
# Save execution trace
trace = {
    "steps": [...],
    "decisions": [...],
    "states": [...]
}
# Later: Replay execution for debugging
```

4. **Deterministic Mode for Testing:**
```python
# Use temperature=0 for reproducible testing
agent = Agent(llm_temperature=0)
```

5. **Bounded Exploration:**
```python
# Limit what paths agent can take
allowed_actions = ["search", "read", "summarize"]
# Agent can only choose from allowed set
```

**The Problem:**

**Layman:** The agent might give you the answer in a completely different format than you expected - sometimes a paragraph, sometimes bullet points, sometimes JSON. It's inconsistent.

**Technical:**
- Output format varies between runs
- Quality inconsistency (sometimes great, sometimes poor)
- May not follow exact requirements
- Hallucinations (making up information)
- Incomplete or over-complete responses

## Guardrails: Essential Safety Measures

**Layman:** Guardrails are like the bumpers in bowling - they keep the ball (agent) from going into the gutter. They're rules and limits that ensure your agent behaves safely and predictably.

**Technical:** Programmatic constraints and validation rules that restrict agent behavior, enforce policies, and ensure outputs meet safety and quality standards.

---

### What are Guardrails?

**Purpose:**
- Prevent harmful actions
- Maintain cost control  
- Ensure consistent behavior
- Build trustworthy systems
- Meet compliance requirements

**Where Guardrails Apply:**
```
Input → [Guardrails] → Agent → [Guardrails] → Output
         ↑                        ↑
    Validate input          Validate actions/output
```

---

### Types of Guardrails:

### 1. Input Validation

**Layman:** Check what's coming IN before the agent sees it - like a security guard checking IDs at the door.

**Purpose:** Ensure inputs are safe, valid, and appropriate

**Examples:**
```python
def validate_input(user_input):
    # Length check
    if len(user_input) > 10000:
        raise ValueError("Input too long")
    
    # Content filtering
    if contains_harmful_content(user_input):
        raise ValueError("Inappropriate content")
    
    # Format validation
    if not is_valid_format(user_input):
        raise ValueError("Invalid format")
    
    return user_input
```

**What to Check:**
- Input length (prevent excessive token usage)
- Prohibited content (profanity, harmful requests)
- Required fields present
- Data types correct
- Within expected ranges

---

### 2. Output Validation

**Layman:** Check what's coming OUT before showing it to users - like quality control on a factory line.

**Purpose:** Ensure outputs are safe, accurate, and meet quality standards

**Examples:**
```python
def validate_output(agent_output):
    # Structure check
    if not has_required_fields(agent_output):
        raise ValueError("Missing required fields")
    
    # Content safety
    if contains_pii(agent_output):
        agent_output = redact_pii(agent_output)
    
    # Factual verification
    if makes_claims(agent_output):
        verify_claims(agent_output)
    
    # Toxicity check
    if is_toxic(agent_output):
        raise ValueError("Toxic content detected")
    
    return agent_output
```

**What to Check:**
- No personal information (PII) leaked
- No toxic/harmful content
- Factual accuracy
- Proper formatting
- Meets quality thresholds

---

### 3. Resource Limits

**Layman:** Set budgets and limits - like giving someone a credit card with a spending limit.

**Purpose:** Prevent runaway costs and resource exhaustion

**Examples:**
```python
class ResourceLimitedAgent:
    def __init__(self):
        self.max_iterations = 15
        self.max_cost = 2.00
        self.max_time = 60  # seconds
        self.max_tokens = 100000
    
    def run(self):
        start_time = time.time()
        
        for i in range(self.max_iterations):
            if self.spent >= self.max_cost:
                break
            
            if time.time() - start_time >= self.max_time:
                break
            
            if self.tokens_used >= self.max_tokens:
                break
            
            self.step()
```

**Resource Limits:**
- Max iterations
- Max cost ($ spent)
- Max time (seconds)
- Max tokens
- Max API calls
- Max file size

### 4. Tool Restrictions

**Layman:** Control which tools the agent can use - like only giving a child access to safety scissors, not sharp knives.

**Purpose:** Prevent agents from taking dangerous or unauthorized actions

### Implementing Guardrails:

**Layered Approach:**
```python
class GuardedAgent:
    def run(self, user_input):
        # Layer 1: Input guardrails
        validated_input = self.input_guardrails(user_input)
        
        # Layer 2: Execution guardrails
        with self.resource_limits():
            with self.tool_restrictions():
                output = self.agent.run(validated_input)
        
        # Layer 3: Output guardrails
        safe_output = self.output_guardrails(output)
        
        return safe_output
```

### Why Guardrails are Critical:

**Without Guardrails:**
- Agents might take harmful actions
- Costs can spiral out of control
- Outputs may be inappropriate
- System behavior unpredictable
- Security vulnerabilities

**With Guardrails:**
- Safe, controlled behavior
- Predictable costs
- Quality assurance
- Compliance with policies
- Production-ready systems

---

**Remember:** Guardrails don't eliminate agent flexibility - they create safe boundaries within which agents can operate autonomously. It's the difference between "do whatever you want" and "do whatever you want within these sensible limits."

**Best Practice:** Start with restrictive guardrails, then relax them as you gain confidence in your agent's behavior!

## Agentic AI Frameworks

**Overview:** Think of these as pre-built toolkits for building AI agents - you could build everything from scratch, but frameworks give you tested patterns and tools to work faster.

**Technical:**
- Direct API calls to LLM providers (OpenAI, Anthropic, etc.)
- Manual implementation of all agent logic
- You write everything: prompts, tools, loops, error handling

- No abstraction layers hiding behavior
- Minimal dependencies
- Best for learning how things work
- Can optimize exactly for your use case

**Cons:**
- Most time-consuming
- Must implement everything yourself
- Easy to make mistakes
- No community patterns/best practices built-in

**Use When:**
- Simple use case that doesn't need complex orchestration
- Learning how agents work under the hood
- Custom requirements that frameworks don't support
- Want absolute control over every API call

```python
import openai
```

### 2. MCP (Model Context Protocol)

**What It Provides:**

## Understanding Asyncio
- All agents frameworks uses asynchronous python
```python

async def do_something() -> str:
    # Do some work
    return 'I am done'

result  = await do_something()
```
- Asyncio provides a lightweight alternative to threading or multiprocessing
- Functions defined with **async def** are called coroutines - they're special functions that can be paused or resumed
- Calling a coroutine doesn't execute it immediately - it returns a coroutine object
- To actually run coroutine, you must **await** it, which schedules it for execution within a **event loop**
- While coroutine is waiting (e.g for I/O), the **event loop** can run the other coroutines

### Core Concepts

**1. Coroutines** - Functions that can pause and resume
```python
async def fetch_data():
    # This is a coroutine
    return "data"
```

**2. Event Loop** - The conductor that manages when coroutines run
```python
# The event loop schedules and runs coroutines
# It's like a task scheduler
```

**3. await** - Pauses execution until the awaited task completes
```python
result = await fetch_data()  # Pause here, let others run
```

---

## Practical Examples

### Example 1: Basic Async/Await

```python
import asyncio

async def make_coffee():
    print("☕ Starting to make coffee...")
    await asyncio.sleep(3)  # Simulates waiting 3 seconds
    print("☕ Coffee is ready!")
    return "Hot Coffee"

async def toast_bread():
    print("🍞 Starting to toast bread...")
    await asyncio.sleep(2)  # Simulates waiting 2 seconds
    print("🍞 Toast is ready!")
    return "Crispy Toast"

# Running ONE task (takes 3 seconds total)
async def make_breakfast_slow():
    coffee = await make_coffee()  # Wait for coffee (3s)
    toast = await toast_bread()   # Then wait for toast (2s)
    print(f"Breakfast: {coffee} and {toast}")
    # Total time: 5 seconds

# Running CONCURRENTLY (takes 3 seconds total)
async def make_breakfast_fast():
    # Start both at the same time
    coffee_task = asyncio.create_task(make_coffee())
    toast_task = asyncio.create_task(toast_bread())
    
    # Wait for both to finish
    coffee = await coffee_task
    toast = await toast_task
    print(f"Breakfast: {coffee} and {toast}")
    # Total time: 3 seconds (max of the two)

# Run it
asyncio.run(make_breakfast_fast())
```

### Example 2: Real-World API Calls

```python
import asyncio
import aiohttp  # Async HTTP library

async def fetch_user(session, user_id):
    print(f"🔍 Fetching user {user_id}...")
    url = f"https://api.example.com/users/{user_id}"
    
    async with session.get(url) as response:
        data = await response.json()
        print(f"✅ Got user {user_id}")
        return data

async def fetch_all_users():
    user_ids = [1, 2, 3, 4, 5]
    
    async with aiohttp.ClientSession() as session:
        # Create tasks for all users
        tasks = [fetch_user(session, uid) for uid in user_ids]
        
        # Wait for all to complete
        results = await asyncio.gather(*tasks)
        return results

# SYNCHRONOUS: Would take 5 seconds (1 second per user)
# ASYNCHRONOUS: Takes ~1 second (all at once)
users = asyncio.run(fetch_all_users())
```

## Key Technical Points

### 1. **Coroutines Don't Run Until Awaited**

```python
async def my_func():
    return "Hello"

# This does NOT run the function!
coro = my_func()  # Returns a coroutine object
print(type(coro))  # <class 'coroutine'>

# This actually runs it:
result = await my_func()  # Inside another async function
# OR
result = asyncio.run(my_func())  # From synchronous code
```

### 2. **The Event Loop**

```python
import asyncio

async def task1():
    print("Task 1 starting")
    await asyncio.sleep(2)
    print("Task 1 done")

async def task2():
    print("Task 2 starting")
    await asyncio.sleep(1)
    print("Task 2 done")

async def main():
    # Event loop will manage these
    await asyncio.gather(task1(), task2())

# Output:
# Task 1 starting
# Task 2 starting
# Task 2 done (after 1 second)
# Task 1 done (after 2 seconds)

asyncio.run(main())
```

**What happens:**
1. Event loop starts task1 → hits `await sleep(2)` → pauses task1
2. Event loop starts task2 → hits `await sleep(1)` → pauses task2
3. After 1 second, task2 resumes and completes
4. After 2 seconds, task1 resumes and completes

### 3. **Why Agent Frameworks Use Asyncio**

```python
# Typical agent workflow
async def agent_process_query(query):
    # These can happen concurrently!
    search_task = asyncio.create_task(web_search(query))
    db_task = asyncio.create_task(database_query(query))
    llm_task = asyncio.create_task(llm_call(query))
    
    # Wait for all results
    search_results, db_results, llm_response = await asyncio.gather(
        search_task, db_task, llm_task
    )
    
    # Combine and return
    return synthesize_results(search_results, db_results, llm_response)
```

**Why this matters:**
- Web searches take 500ms
- Database queries take 300ms
- LLM calls take 2 seconds
- **Synchronous:** 2.8 seconds total
- **Asynchronous:** 2 seconds total (the longest task)

## Common Patterns

### Pattern 1: Running Multiple Tasks
```python
async def main():
    # Method 1: gather
    results = await asyncio.gather(task1(), task2(), task3())
    
    # Method 2: create_task
    t1 = asyncio.create_task(task1())
    t2 = asyncio.create_task(task2())
    await t1
    await t2
```

### Pattern 2: Timeout
```python
async def slow_operation():
    await asyncio.sleep(10)
    return "done"

try:
    result = await asyncio.wait_for(slow_operation(), timeout=5)
except asyncio.TimeoutError:
    print("Operation took too long!")
```

### Pattern 3: Running Sync Code in Async
```python
import asyncio

def blocking_function():
    # Some CPU-intensive work
    return sum(range(1000000))

async def main():
    # Run blocking code in thread pool
    result = await asyncio.to_thread(blocking_function)
```

## When to Use Asyncio

**✅ Great for:**
- Network I/O (API calls, web scraping)
- Database queries
- File I/O
- Multiple LLM calls
- Agent frameworks with multiple tools

**❌ Not ideal for:**
- CPU-intensive calculations (use multiprocessing)
- Code with lots of synchronous libraries
- Simple scripts that don't wait for anything

The key insight: asyncio shines when you're **waiting** for external resources, not when you're **computing**.

## Vibe Coding - AI-Assisted Development Philosophy

**Layman:** "Vibe coding" is a playful term for using AI coding assistants effectively. Instead of worrying about perfection, you "vibe" with the AI to quickly generate code, then verify and refine it.

**Technical:** A pragmatic approach to AI-assisted development that emphasizes rapid iteration, validation, and multiple perspectives over perfect first-attempts.

---

### The 5 Vibe Coding Principles:

### 1. Good Vibe - Prompt Well

**Concept:** The quality of AI output depends heavily on your prompt quality.

**Best Practices:**
- **Be specific:** Not "make a function" but "write a Python function that validates email addresses using regex"
- **Ask for short answers:** "Give me a concise solution" prevents verbose output
- **Request latest APIs:** "Use the latest OpenAI API as of Oct 2025" ensures current practices
- **Specify style:** "Write this in TypeScript with async/await"

**Example:**
```
❌ Bad: "Write code for authentication"

✅ Good: "Write a Python function for JWT authentication using the latest
PyJWT library. Include token generation, validation, and expiration
handling. Keep it under 50 lines. Add type hints."
```

**Why it matters:**
- Better prompts = better code
- Saves iteration time
- Gets you closer to what you actually need

### 2. Vibe and Verify - Ask 2 LLMs

**Concept:** Different LLMs have different strengths. Ask the same question to multiple LLMs and compare answers.

**How to do it:**


**What to look for:**

**Why it matters:**

### 5. Vibe with Variety - Ask for 3 Solutions

**Concept:** Request multiple approaches to the same problem, then pick the best one.

**How to prompt:**
```
"Provide 3 different solutions for implementing caching in my API:
1. A simple in-memory solution
2. A Redis-based solution
3. A database-backed solution

For each, briefly explain:
- Pros and cons
- When to use it
- Complexity level"
```

**AI might provide:**

**Solution 1: In-Memory (Dict)**
```python
cache = {}
def get_data(key):
    if key in cache:
        return cache[key]
    data = expensive_operation(key)
    cache[key] = data
    return data
```
- Pros: Simple, fast
- Cons: Lost on restart, not shared across servers
- Use: Single server, simple caching needs

**Solution 2: Redis**
```python
import redis
r = redis.Redis()

def get_data(key):
    cached = r.get(key)
    if cached:
        return cached
    data = expensive_operation(key)
    r.setex(key, 3600, data)
    return data
```
- Pros: Persistent, distributed, feature-rich
- Cons: External dependency, more complex
- Use: Multi-server, production apps

**Solution 3: Database-backed**
```python
def get_data(key):
    cached = db.query(Cache).filter_by(key=key).first()
    if cached and not_expired(cached):
        return cached.value
    data = expensive_operation(key)
    db.add(Cache(key=key, value=data))
    return data
```
- Pros: No new infrastructure, persistent
- Cons: Slower than Redis, more DB load
- Use: Already have DB, don't want Redis

**Now pick the best for YOUR situation!**

**Why it matters:**
- See trade-offs clearly
- Understand different approaches
- Make informed decisions
- Learn multiple patterns

## OpenAI Agents SDK - Advanced Concepts

### The Progression of Agent Complexity:

**Layman:** Start simple, add complexity only when needed. Here's the natural progression:

1. **Single agent workflow** - One AI doing one task
2. **Agent with tools** - AI that can use functions  
3. **Multi-agent with handoffs** - Multiple AIs collaborating

---

### 1. A Workflow of Agent Calls

**Concept:** Sequential agent invocations where each agent does a specific task.

**Example - Content Pipeline:**
```python
from agents import Agent, Runner

# Define specialized agents
researcher = Agent(
    name="Researcher",
    instructions="Research topics and gather facts"
)

writer = Agent(
    name="Writer",
    instructions="Write engaging content based on research"
)

editor = Agent(
    name="Editor",
    instructions="Edit content for grammar and clarity"
)

# Workflow: Research → Write → Edit
async def content_workflow(topic):
    # Step 1: Research
    research = await Runner.run(researcher, f"Research {topic}")
    
    # Step 2: Write using research
    draft = await Runner.run(writer, f"Write article based on: {research.final_output}")
    
    # Step 3: Edit
    final = await Runner.run(editor, f"Edit this: {draft.final_output}")
    
    return final.final_output
```

**Key Point:** Each agent runs independently. You manually orchestrate the flow.

---

### 2. An Agent That Can Use Tools

**Concept:** Give an agent functions it can call to take actions or retrieve information.

**Example - Research Assistant with Tools:**
```python
from agents import Agent, Runner

def search_web(query: str) -> str:
    """Search the web for information"""
    # Implementation here
    return f"Search results for: {query}"

def fetch_stock_price(symbol: str) -> str:
    """Get current stock price"""
    # Implementation here
    return f"Stock {symbol}: $150.00"

# Agent with tools
research_agent = Agent(
    name="Research Assistant",
    instructions="Help users research topics. Use tools when needed.",
    functions=[search_web, fetch_stock_price]
)

# Agent will automatically use tools when appropriate
result = await Runner.run(
    research_agent,
    "What's the current price of AAPL stock?"
)
# Agent will call fetch_stock_price("AAPL") automatically
```

**What Happens:**
1. Agent receives question
2. Agent decides to use `fetch_stock_price` tool
3. SDK executes the function
4. Agent uses result to form response

**Key Point:** Agent autonomously decides when to use tools.

---

### 3. An Agent That Can Call Other Agents

**Now the interesting part:** Agents can delegate to other agents!

---

## Tools vs Handoffs - Critical Distinction

**Layman:** This is one of the most important concepts to understand!

### Tools = Temporary Delegation

**How it works:**
```
Agent A → Calls Tool → Tool Returns Result → Agent A Continues
         "Borrow"              "Return"        "Still in control"
```

**Example:**
```python
agent = Agent(
    name="Main",
    instructions="Help users",
    functions=[search_web]  # Tool
)

# Flow:
# User: "Search for Python tutorials"
# Agent: *calls search_web*
# Tool: *returns results*
# Agent: *uses results to respond*
# Control returns to Agent
```

**Characteristics:**
- Agent calls function
- Function returns value
- **Original agent keeps control**
- Like calling a function in code

---

### Handoffs = Permanent Delegation

**How it works:**
```
Agent A → Hands Off to Agent B → Agent B Takes Over → Agent B Responds
        "You take it"             "I'm in charge now"
```

**Example:**
```python
sales_agent = Agent(
    name="Sales",
    instructions="Handle sales questions"
)

support_agent = Agent(
    name="Support",
    instructions="Handle technical support"
)

triage_agent = Agent(
    name="Triage",
    instructions="Route users to appropriate agent",
    handoffs=[sales_agent, support_agent]  # Can transfer control
)

# Flow:
# User: "I have a technical problem"
# Triage Agent: *decides support is needed*
# Triage Agent: *hands off to Support Agent*
# Support Agent: *now handles conversation*
# Triage Agent: *no longer involved*
```

**Characteristics:**
- Agent transfers control
- New agent takes over
- **Original agent doesn't get control back**
- Like calling another person, not a function

---

### The Key Difference:

| Tools | Handoffs |
|-------|----------|
| Borrow → Return | Pass → Gone |
| Get result back | Don't get control back |
| Same agent continues | Different agent continues |
| Like function call | Like transfer call |
| Agent uses tool's output | New agent takes over |

---

### Visual Comparison:

**With Tools:**
```
User → Agent A → [Call Tool] → [Tool Returns] → Agent A Responds → User
                     ↓              ↑
                     └──────────────┘
                   (Round trip)
```

**With Handoffs:**
```
User → Agent A → [Handoff to Agent B] → Agent B Responds → User
                                         (Agent A done)
```

---

### When to Use Each:

**Use Tools When:**
- Need to fetch data
- Perform calculation
- Execute action
- Want same agent to use result

**Example:** Weather agent calls `get_weather` tool, uses result to respond

**Use Handoffs When:**
- Different expertise needed
- Context switch required
- Specialist agent better suited
- Want different agent to continue

**Example:** General agent hands off to "Python Expert" agent for code questions

---

### Real-World Example:

**Customer Service System:**


### Code Example - Both Tools and Handoffs:

```python

    """Evaluate mathematical expression"""
```

### Remember:

**Tools:**
- Control passes back
- Agent uses result
- Synchronous
- Function-like

**Handoffs:**
- Control passes across
- Agent delegates completely
- Transfer-like
- Conversation continues with new agent

**Both are powerful - use the right one for your use case!**

### 3. Guardrails in OpenAI Agents SDK

**Layman:** Guardrails are safety checks on what your agents can do - like child safety locks but for AI.

**Key Concept:** Guardrails themselves can be agents!

**Important Limitation:** Guardrails can only be applied to:
1. **Input** of the very first agent (what goes in)
2. **Output** of the last agent (what comes out)

You **cannot** insert guardrails in the middle of agent workflows.

---

### Guardrail Types:

**A. Input Guardrails - Check Before Processing**

```python
from agents import Agent, Runner

# Guardrail agent for input validation
input_guard = Agent(
    name="Input Validator",
    instructions="""
    Check if input is appropriate. Reject if:
    - Contains profanity
    - Asks for illegal activities
    - Is abusive or harmful
    
    Return 'APPROVED' or 'REJECTED: reason'
    """
)

# Main agent
assistant = Agent(
    name="Assistant",
    instructions="Help users with their questions"
)

# Apply input guardrail
async def safe_run(user_input):
    # 1. Check input first
    guard_result = await Runner.run(input_guard, user_input)
    
    if "REJECTED" in guard_result.final_output:
        return "I cannot process that request."
    
    # 2. If approved, run main agent
    result = await Runner.run(assistant, user_input)
    return result.final_output
```

---

**B. Output Guardrails - Check Before Returning**

```python
# Guardrail agent for output validation
output_guard = Agent(
    name="Output Validator",
    instructions="""
    Check if response is appropriate. Flag if:
    - Contains personal information (PII)
    - Has toxic content
    - Includes misinformation
    
    Return 'SAFE' or 'UNSAFE: reason'
    """
)

async def safe_output(user_input):
    # 1. Generate response
    result = await Runner.run(assistant, user_input)
    
    # 2. Validate output
    guard_result = await Runner.run(
        output_guard,
        f"Check this response: {result.final_output}"
    )
    
    if "UNSAFE" in guard_result.final_output:
        return "I apologize, but I cannot provide that response."
    
    return result.final_output
```

---

**C. Combined Input + Output Guardrails**

```python
async def fully_guarded_agent(user_input):
    # Input guardrail
    input_check = await Runner.run(input_guard, user_input)
    if "REJECTED" in input_check.final_output:
        return "Request rejected."
    
    # Main processing
    result = await Runner.run(assistant, user_input)
    
    # Output guardrail
    output_check = await Runner.run(output_guard, result.final_output)
    if "UNSAFE" in output_check.final_output:
        return "Response blocked for safety."
    
    return result.final_output
```

---

**D. Content Filtering Guardrail**

```python
from pydantic import BaseModel

class ContentCheck(BaseModel):
    is_appropriate: bool
    reason: str
    severity: str  # "low", "medium", "high"

content_filter = Agent(
    name="Content Filter",
    instructions="Analyze content for appropriateness",
    response_format=ContentCheck
)

async def filtered_response(user_input):
    # Check input
    check = await Runner.run(content_filter, user_input)
    content_check = check.structured_output
    
    if not content_check.is_appropriate:
        if content_check.severity == "high":
            log_violation(user_input, content_check.reason)
            return "Request blocked."
    
    # Process if appropriate
    result = await Runner.run(assistant, user_input)
    return result.final_output
```

---

### Guardrail Limitations:

**What You CAN Do:**
```
[Input Guardrail] → Agent Chain → [Output Guardrail]
      ✅                              ✅
```

**What You CANNOT Do:**
```
Agent A → [Guardrail] → Agent B → [Guardrail] → Agent C
              ❌                       ❌
```

**Why:** Agents SDK doesn't support mid-workflow guardrails. You'd need to implement this manually.

---

### Guardrail Best Practices:

**1. Keep Guardrails Simple**
- Fast evaluation (don't slow down responses)
- Clear accept/reject criteria
- Use cheaper models when possible

**2. Log Everything**
```python
if input_rejected:
    logger.warning(f"Input rejected: {user_input} - Reason: {reason}")
```

**3. Multiple Layers**
- Input validation
- Output validation  
- Content filtering
- Rate limiting

**4. Clear User Feedback**
```python
if rejected:
    return "I cannot help with that. If you think this is an error, please contact support."
```

**5. Monitor and Improve**
- Track rejection rates
- Review false positives
- Refine guardrail prompts

## MCP (Model Context Protocol) - Understanding the Hype

**Created by:** Anthropic (makers of Claude)

---

### What MCP Is NOT:

**Layman:** Before getting excited, understand what MCP ISN'T - this prevents disappointment!

**1. NOT a Framework for Building Agents**
- Doesn't replace LangChain, AutoGen, or OpenAI SDK
- Doesn't orchestrate agent workflows
- Doesn't manage agent conversations
- Just handles connections between apps and tools

**2. NOT a Fundamental Change to How Agents Work**
- Agents still work the same way (LLM + tools)
- Doesn't make agents smarter or better
- Doesn't change the underlying mechanics
- Just standardizes the interface

**3. NOT a Way to Code Agents**
- You still write agents using frameworks (OpenAI SDK, LangChain, etc.)
- MCP just provides the tools your agent can use
- Your agent logic remains the same

---

### What MCP Actually IS:

**Layman:** MCP is a **standard** - like how USB-C is a standard for connecting devices.

**Technical:** A protocol specification for standardizing how AI applications connect to external tools, data sources, and services.

---

### The USB-C Analogy:

**Before USB-C:**
- Every device had different ports
- Need adapters for everything
- Frustrating and messy
- Each vendor did their own thing

**With USB-C:**
- One standard port
- Everything connects the same way
- Works across devices
- Plug and play

**Before MCP:**
- Every agent framework has custom tool format
- Each app implements tools differently
- Tedious to integrate tools
- Reinvent the wheel every time

**With MCP:**
- Standard protocol for tools/resources
- Works across AI applications
- Tool servers can be reused
- Write once, use everywhere

---

### Simple Explanation:

**MCP is like having a standard electrical outlet:**
- Any appliance (AI app) can plug in
- Any device (tool server) that follows the standard works
- Don't need custom adapters for each combination

**Without MCP:**
```
Claude app → Custom integration → Notion
Claude app → Custom integration → Google Drive
Claude app → Custom integration → Slack
ChatGPT app → Different custom integration → Notion
...each app × each service = many integrations
```

**With MCP:**
```
Claude app → MCP → MCP Server (Notion, Google Drive, Slack, etc.)
ChatGPT app → MCP → Same MCP Servers work!
Any MCP app → MCP → Any MCP Server
```

## MCP Core Concepts - The Three Components

**Overview:** MCP has a three-tier architecture: Hosts, Clients, and Servers. Think of it like a phone call system - you have the phone (host), the connection (client), and the person you're calling (server).

---

### 1. Hosts

**Layman:** The host is the AI application you're actually using - like Claude Desktop, your custom agent app, or any AI chatbot.

**Technical:** The host is the LLM application that wants to use external tools and resources. It contains the user interface and orchestrates the overall interaction.

**Examples of Hosts:**
- **Claude Desktop** (Anthropic's desktop app)
- **Your custom agent** (built with OpenAI SDK, LangChain, etc.)
- **VS Code with AI extensions**
- **Any LLM application** that supports MCP

**What Hosts Do:**
- Run the user interface
- Manage conversation flow
- Contain MCP clients
- Display results to users
- Handle user authentication

---

### 2. Clients

**Layman:** The client is like a translator that lives inside your AI app. It speaks to MCP servers on behalf of your app.

**Technical:** MCP clients are embedded within hosts and handle the protocol-level communication with MCP servers. There's a 1:1 relationship - each client connects to exactly one server.

**Key Points:**
- **Lives inside the host** (not a separate program)
- **1:1 connection** - One client per server
- **Handles communication** - Translates between host and server
- **Protocol implementation** - Manages MCP protocol details

**What Clients Do:**
- Establish connections to servers
- Send requests (tool calls, resource requests)
- Receive responses
- Handle errors and retries
- Manage connection lifecycle

**Example:**
```
Your Agent App (Host)
    ├── MCP Client 1 → Connects to Notion MCP Server
    ├── MCP Client 2 → Connects to Google Drive MCP Server
    └── MCP Client 3 → Connects to Database MCP Server
```

Each client manages one connection.

---

### 3. Servers

**Layman:** Servers are the actual tools - they do the real work like searching the web, reading files, querying databases, etc.

**Technical:** MCP servers expose tools, resources, and prompts that hosts can use. They implement the actual functionality and run as separate processes.

**What Servers Provide:**
- **Tools** - Functions the LLM can call (e.g., search, database query)
- **Resources** - Data/context (e.g., files, documents)
- **Prompts** - Reusable prompt templates

**Examples of MCP Servers:**
- **Fetch** - Web searching via headless browser
- **Postgres** - Database queries
- **Filesystem** - File operations
- **Slack** - Messaging integration
- **GitHub** - Repository operations
- **Notion** - Knowledge base access

---

### Real-World Example: Claude Desktop + Fetch

**Scenario:** Using Claude Desktop to search the web

**The Flow:**

1. **Host (Claude Desktop)**
   - You type "What's the latest news on AI?"
   - Claude Desktop is the host application

2. **Client (Inside Claude Desktop)**
   - Claude Desktop has an MCP client
   - Client connects to Fetch MCP Server

3. **Server (Fetch MCP Server)**
   - Fetch server runs on your computer
   - Receives search request from client
   - Uses headless browser to search
   - Returns results

4. **Response Flow:**
   - Fetch → Client → Claude Desktop → You see results

**Most Common: Local (Your Computer)**

**Layman:** MCP servers usually run on your own computer, not in the cloud. You download them and run them locally.

**Why Local?**
- **Privacy** - Your data stays on your machine
- **Security** - No external access needed
- **Control** - You manage what runs
- **Free** - No server hosting costs

**How It Works:**
```bash

# Server runs on your machine
# Client connects via localhost
```

**Alternative: Remote Servers**
- Can run on remote servers via HTTPS
- Less common but possible
- Useful for shared resources

## Two Transport Mechanisms:

### 1. Stdio (Standard Input/Output)

**Layman:** Like typing commands into a program's terminal - you send text in, get text back.

**Technical:**
- Spawns MCP server as a child process
- Communication via stdin/stdout
- Server runs on same machine as host
- Simplest transport mechanism

**When to Use:**
- Local development
- Single-user applications
- Simple tool integrations
- Desktop applications

**Example Configuration:**
```python
fetch_params = {
    "command": "uvx",  # Command to run
    "args": ["mcp-server-fetch"]  # Server to launch
}
```

**How it works:**
```
Host → Launches Server Process
    ↓
Host writes to stdin → Server reads stdin
Server writes to stdout → Host reads stdout
```

---

### 2. SSE (Server-Side Events)

**Layman:** Like a phone call over the internet - maintains an open connection for back-and-forth communication.

**Technical:**
- Uses HTTPS connection with streaming
- Server can be remote
- Supports real-time updates
- More scalable for multi-user scenarios

**When to Use:**
- Remote servers
- Web applications
- Multi-user systems
- Need real-time updates

**Example:**
```
Host → HTTPS connection → Remote MCP Server
    ↑                           ↓
    └──── Streaming events ─────┘
```

**Comparison:**

| Feature | Stdio | SSE |
|---------|-------|-----|
| **Location** | Local only | Local or remote |
| **Connection** | Process spawn | HTTPS |
| **Use Case** | Desktop apps | Web apps |
| **Complexity** | Simpler | More complex |
| **Scaling** | Single user | Multi-user |

### Step 4: Use in Your Code

**Python Example with MCP:**
```python
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client

# Define server parameters
server_params = StdioServerParameters(
    command="uvx",
    args=["mcp-server-fetch"]
)

# Connect to server
async with stdio_client(server_params) as (read, write):
    async with ClientSession(read, write) as session:
        # Initialize connection
        await session.initialize()
        
        # List available tools
        tools = await session.list_tools()
        
        # Call a tool
        result = await session.call_tool("fetch", {
            "url": "https://example.com"
        })
```
