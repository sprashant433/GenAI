### 🧠 Context Window

**The maximum number of tokens a model can remember at once.**

Think of it as the model's **short-term memory** or a **whiteboard with limited space**.

**Key Points:**
- Includes **input + output** tokens combined
- When limit is exceeded, oldest parts are "forgotten"
- Larger context = can handle longer documents/conversations

**Model Limits:**
| Model | Token Limit |
|-------|-------------|
| GPT-3.5 | ~4,096 |
| GPT-4 | ~8,192 to 128,000 |
| Claude 3.5 | ~200,000+ |
| Gemini 1.5 | 1 million |

## Chat Models & Instruct Variants

### What is an "Instruct" model?

> Models that have been **trained to follow instructions in a conversational format** – like ChatGPT.

#### Naming:

* These models often end in **`-instruct`**, like:

  * `"mistral-7b-instruct"`
  * `"llama-3-instruct"`

---

### Format of Instruct Prompt

Instruct models expect **chat-style messages**:

```python
[
  {"role": "system", "content": "You are a helpful assistant."},
  {"role": "user", "content": "What's the weather today?"}
]
```

#### `apply_chat_template()` – Your Prompt Formatter

> Think of this as a **"message packager"** – it converts your list of chat messages into the **exact string format** the model expects.

#### 🧪 Example:

````python
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B-Instruct")

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Tell me a joke."}
]

prompt = tokenizer.apply_chat_template(messages, tokenize=False)
print(prompt)
# <s>[INST] <<SYS>>You are a helpful assistant.<</SYS>> Tell me a joke. [/INST]
````

## **LLM Message Format (Chat Style)**

### What?

* Conversations are structured as a **list of messages** with roles.

### Format:

```python
[
    {"role": "system", "content": "Defines assistant behavior"},
    {"role": "user", "content": "User message"},
    {"role": "assistant", "content": "Assistant reply"}
]
```


### Roles:

* **system**: Sets assistant behavior (e.g., tone, expertise)
* **user**: Asks questions or gives prompts
* **assistant**: Responds to user inputs

### Why?

* Maintains **context**
* Enables **multi-turn chat**
* Helps in **fine-tuning behavior**

### Three Main Access Methods

#### 1. **Chat Interface** (Easiest)
- **Examples:** ChatGPT, Claude, Gemini
- **Best for:** Non-coders, quick Q&A, brainstorming
- **How:** Type questions, get conversational responses

#### 2. **Cloud APIs** (For Developers)
**Options:**
- **Direct APIs:** OpenAI API, Anthropic API
- **Frameworks:** LangChain for complex applications
- **Managed Services:** AWS Bedrock, Google Vertex AI, Azure ML

**Best for:** Building apps, integrating AI into products

````python
from google.colab import userdata
from openai import OpenAI
openai_api_key = userdata.get('OPENAI_API_KEY')
openai_client = OpenAI(api_key = openai_api_key)

system_message = "You are an assistant that is great at telling jokes"
user_prompt = "Tell a light-hearted joke for an audience of Data Scientists"
prompts = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
  ]

completion = openai_client.chat.completions.create(
    model='gpt-4o-mini',
    messages=prompts,
    temperature=0.7 # Temperature setting controls creativity
)
print(completion.choices[0].message.content)
````


## **Streaming**

Normally, when you ask an LLM something, it **cooks the whole reply first** and **then serves** it (like a waiter bringing the entire pizza).

**Streaming** is when the model **sends tokens as soon as they’re ready** — so you can start reading before the full text is generated.

* Helps in **real-time apps** (chatbots, transcription, coding assistants).
* Feels **instant** even for long outputs.

- Instead of waiting for complete response, tokens arrive one-by-one (typewriter effect).

**Example:**

````python
from google.colab import userdata
openai_api_key = userdata.get('OPENAI_API_KEY')

from openai import OpenAI
openai = OpenAI(api_key = openai_api_key)

system_message = "You are an assistant that is great at telling jokes"
user_prompt = "Tell a light-hearted joke for an audience of Data Scientists"
prompts = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
  ]

stream = openai.chat.completions.create(
    model='gpt-4o-mini', 
    messages=prompts,
    temperature=0.7,
    stream=True,
)

for chunks in stream:
  print(chunks.choices[0].delta.content or "", end="")
````


**Analogy:**

* Non-streaming: You wait for the chef to bake the whole cake.
* Streaming: You get each slice as soon as it’s ready.

## Several ways to achieve the same functionality using LangChain:

## Method 1: Using ChatOpenAI (Most Common)

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# Initialize the model
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7,
    openai_api_key=openai_api_key
)

# Create messages
messages = [
    SystemMessage(content="You are an assistant that is great at telling jokes"),
    HumanMessage(content="Tell a light-hearted joke for an audience of Data Scientists")
]

# Get response
response = llm.invoke(messages)
print(response.content)
````


## Method 2: Using Chat Prompt Templates

````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Initialize model
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7,
    openai_api_key=openai_api_key
)

# Create prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an assistant that is great at telling jokes"),
    ("human", "{user_input}")
])

# Create chain
chain = prompt | llm

# Execute
response = chain.invoke({"user_input": "Tell a light-hearted joke for an audience of Data Scientists"})
print(response.content)
````


## Method 3: Using LLMChain (Legacy but still works)

````python
from langchain_openai import ChatOpenAI
from langchain.chains import LLMChain
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7,
    openai_api_key=openai_api_key
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an assistant that is great at telling jokes"),
    ("human", "{user_input}")
])

chain = LLMChain(llm=llm, prompt=prompt)
response = chain.run(user_input="Tell a light-hearted joke for an audience of Data Scientists")
print(response)
````


## Method 4: Using LCEL (LangChain Expression Language) - Most Flexible

````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Initialize components
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7,
    openai_api_key=openai_api_key
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an assistant that is great at telling jokes"),
    ("human", "{user_input}")
])

output_parser = StrOutputParser()

# Create chain using LCEL
chain = prompt | llm | output_parser

# Execute
response = chain.invoke({"user_input": "Tell a light-hearted joke for an audience of Data Scientists"})
print(response)
````


## Method 5: With Other Providers (Anthropic, Google, etc.)

````python
# For Anthropic Claude
from langchain_anthropic import ChatAnthropic

llm = ChatAnthropic(
    model="claude-3-5-sonnet-20241022",
    temperature=0.7,
    anthropic_api_key=anthropic_api_key
)

# For Google Gemini
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-pro",
    temperature=0.7,
    google_api_key=google_api_key
)

# Same usage pattern with any provider
messages = [
    SystemMessage(content="You are an assistant that is great at telling jokes"),
    HumanMessage(content="Tell a light-hearted joke for an audience of Data Scientists")
]
response = llm.invoke(messages)
print(response.content)
````


**Key Differences:**
- **Method 1**: Direct equivalent, most straightforward
- **Method 2**: More reusable with templates
- **Method 3**: Legacy approach, still functional
- **Method 4**: Most flexible, allows easy chaining with other components
- **Method 5**: Same pattern works across different LLM providers

**Recommended**: Use Method 2 or 4 for most applications as they're more maintainable and follow LangChain's current best practices.

## Limitations & Challenges

### Key Limitations

1. **Knowledge Cutoff**
   - No recent events unless web-connected
   - Training data has fixed cut-off date

2. **Hallucinations**
   - Can confidently state false information
   - May invent citations, APIs, or facts

3. **Reasoning Gaps**
   - Mimics reasoning but fails complex logic
   - Struggles with multi-step problems

4. **Domain Expertise**
   - Broad but shallow knowledge
   - Not truly expert-level in specialized fields

5. **Bias & Safety**
   - Reflects training data biases
   - Can generate harmful content if not properly aligned

6. **Resource Intensive**
   - High computational costs
   - Environmental impact concerns

7. **Context Limitations**
   - Fixed token limits
   - Forgets information beyond window

8. **Inconsistency**
   - Same prompt can yield different outputs
   - Probabilistic, not deterministic
